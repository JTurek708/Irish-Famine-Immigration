## Purpose

This notebook will serve as an aggregation notebook, bringing in the data, joining the data on different fields, and creating an aggregation data frame that can be used for further analysis

```python
import pandas as pd
```

```python
# Bring in data 
country_codes = pd.read_csv('country_codes.csv')
famine_ships = pd.read_csv('famine_Irish_Ships.csv')
job_codes = pd.read_csv('jobs_codes.csv')
port_codes = pd.read_csv('port_codes.csv')
```

```python
fipas = pd.read_csv('fipas.csv')

fipas.head()
```

```python
job_codes.head()
```

```python
# Merge fipas with job_codes
# Create new df 'agg_data'
agg_data = pd.merge(fipas, job_codes, how='left',
                    left_on='occupation_code',
                    right_on='job_code')

print(agg_data.head())
```

```python
print(country_codes.tail(10))
```

```python
# Change country_code column to integer
agg_data['native_country'] = agg_data['native_country'].apply(lambda x: int(x) if pd.notnull(x) else x)
print(agg_data.head())
```

```python
null_count=agg_data['native_country'].isnull().sum()
print(null_count)
```

```python
# Replace nulls with '99'
agg_data['native_country'] = agg_data['native_country'].fillna(99)
```

```python
# Confirm that native_country values are int
agg_data['native_country'] = agg_data['native_country'].astype(int)
```

```python
# Confirm that country_code in country_codes is int
country_codes['country_code'] = country_codes['country_code'].astype(int)
```

```python
# Merge agg_data with country_codes
agg_data = pd.merge(agg_data, country_codes, how='left',
                    left_on='native_country',
                    right_on='country_code')

print(agg_data)
```
